## Q1: Sorting Hat
In this problem, I created a function called sorting_hat that asks the user a few yes/no questions. Depending on the answers, the function uses if/elif/else logic to decide whether the user belongs to Gryffindor or Slytherine. The function then returns the assigned house.

```python
def sorting_hat():
    q1 = input("Do you value bravery, courage, and chivalry? (Y/N): ").strip().upper()
    q2 = input("Do you value ambition, resourcefulness, and self-preservation (Y/N): ").strip().upper()

    if q1 == 'Y' and q2 == 'N':
        house = "Gryffindor"
    elif q1 == 'N' and q2 == 'Y':
        house = "Slytherine"
    elif q1 == 'Y' and q2 == 'Y':
        q3 = input("When faced with a high-stakes, dangerous situation, is your immediate instinct to charge in for the sake of doing what is right (Y), or to pause and calculate how to win while ensuring your own survival (N)? (Y/N): ").strip().upper()
        if q3 == 'Y':
            house = "Gryffindor"
        else:            house = "Slytherine"
    else:
        house = "Gryffindor"
    return house
sorting_hat()
```

## Q2: mydata class
In this problem, I created a class called mydata that takes in a list of numbers. The class includes two methods (mean and std). I then tested the class using a sample list.

```python
class mydata:
    def __init__(self, data):
        self.data = data

    def mean(self):
        return sum(self.data) / len(self.data)

    def std(self):
        mu = self.mean()
        variance = sum((x - mu) ** 2 for x in self.data) / len(self.data)
        return variance ** 0.5

x = mydata([1, 2, 7, 5])
print("mean =", x.mean())
print("std =", x.std())
```

## Q3: lag function
In this problem, I created a function called lag that takes in a numpy array and an integer n. The function returns a lagged version of the array by shifting the values to the right and filling the first n positions with numpy.nan.

```python
import numpy as np

def lag(a, n):
    result = np.empty(len(a), dtype=float)
    result[:n] = np.nan
    result[n:] = a[:-n]
    return result

a = np.array([1, 6, 8, 4])
print(lag(a, 1))
print(lag(a, 2))
```

## Q4: lagGenerator function
In this problem, I created a function called lagGenerator that takes an integer n and returns a new function. The returned function uses a lambda expression to apply the lag function from Q3 with the specified lag value.

```python
def lagGenerator(n):
    return lambda a: lag(a, n)
lag2 = lagGenerator(2)
print(lag2(np.array([1, 6, 8, 4])))
```

## Q5: filter_and_square function
In this problem, I wrote a function called filter_and_square. Using a list comprehension, the function returns a new list containing the squares of only the positive numbers from the original list.

```python
def filter_and_square(nums):
    return [x ** 2 for x in nums if x > 0]
print(filter_and_square([-2, 0, 3, 4]))
```

## Q6: frequency_map function
In this problem, I created a function that counts the frequency of each string in a list using a dictionary. Its count is stored as the value.

```python
def frequency_map(strings):
    freq = {}
    for s in strings:
        if s in freq:
            freq[s] += 1
        else:
            freq[s] = 1
    return freq
print(frequency_map(['apple', 'banana', 'apple', 'orange']))
```

## Q7: extended_data class
In this problem, I created a class called extended_data that inherits from the mydata class. In addition to the mean and std methods, I added min and max, which return the minimum and maximum values in the data.

```python
class extended_data(mydata):
    def min(self):
        return min(self.data)

    def max(self):
        return max(self.data)
y = extended_data([1, 2, 7, 5])
print("mean =", y.mean())
print("std =", y.std())
print("min =", y.min())
print("max =", y.max())
```
